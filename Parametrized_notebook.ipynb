```python
dbutils.widgets.text("process_date", "")
dbutils.widgets.text("txn_type", "")
```

```python
process_date = dbutils.widgets.get("process_date")
txn_type = dbutils.widgets.get("txn_type")

print("Process Date:", process_date)
print("Transaction Type:", txn_type)
```

```python
from pyspark.sql import functions as F

df = spark.table("card_transactions")
```

```python
result_df = (
    df
    .filter(F.col("txn_date") == process_date)
    .filter(F.col("txn_type") == txn_type)
    .filter(F.col("status") == "APPROVED")
)
display(result_df)
```
