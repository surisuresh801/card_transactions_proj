```python
from pyspark.sql import functions as F

df = spark.table("card_transactions")

invalid_df = df.filter(
    F.col("txn_id").isNull() |
    F.col("amount").isNull() |
    (F.col("amount") <= 0)
)

invalid_count = invalid_df.count()

print("Invalid records:", invalid_count)

if invalid_count > 0:
    raise Exception(
        f"Data quality failed. Invalid records: {invalid_count}"
    )

print("Data quality validation successful.")

# raise Exception("Testing job failure and retry")

```
