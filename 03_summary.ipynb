```python
from pyspark.sql import functions as F

dbutils.widgets.text("process_date", "")
dbutils.widgets.text("txn_type", "")

process_date = dbutils.widgets.get("process_date")
txn_type = dbutils.widgets.get("txn_type")

df = spark.table("card_transactions")

summary_df = (
    df
    .filter(
        (F.col("txn_date") == process_date) &
        (F.col("txn_type") == txn_type)
    )
    .groupBy("txn_date", "txn_type")
    .agg(
        F.count("*").alias("txn_count"),
        F.sum("amount").alias("total_amount")
    )
)

display(summary_df)
```
